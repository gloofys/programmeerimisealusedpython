```python
import timeit
import random
import tracemalloc
```

```python
RANDOM_SEED = 42
TIMING_REPEATS = 5
NUMBER_REPEATS = 10
```

```python
random.seed(RANDOM_SEED)
```

```python
def u1_duplicated(numbers:list)->bool:
    n=len(numbers)
    for i in range(n):
        for j in range(i+1,n):
            if numbers[i]==numbers[j]:
                return True
    return False

def u2_duplicated(numbers:list)->bool:
    numbers_set=set(numbers)
    if len(numbers_set)==len(numbers):
        return False
    return True
```

```python
sizes = [10, 100, 1000, 5_000, 10_000]
```

```python
def measure_time(function, numbers, number = NUMBER_REPEATS, repeat=TIMING_REPEATS):
    results = timeit.repeat(
        lambda: function(numbers),
        number=number,
        repeat=repeat
    )
    return min(results)
```

```python


#parem

print("BEST")
for n in sizes:
    numbers_best = [1,1] + [k for k in range(2, n-1)]
    print(f'1. n = {n} and results = {measure_time(u1_duplicated, numbers_best)}')
    print(f'2. n = {n} and results = {measure_time(u2_duplicated, numbers_best)}')



#average
print("AVERAGE")
for n in sizes:
    numbers_average = [random.randint(1, n) for _ in range(n)]
    print(f'1. n = {n} and results = {measure_time(u1_duplicated, numbers_average)}')
    print(f'2. n = {n} and results = {measure_time(u2_duplicated, numbers_average)}')
    

#halvim
print("WORST")
for n in sizes:
    numbers_worst = [k for k in range(n)]
    print(f'1. n = {n} and results = {measure_time(u1_duplicated, numbers_worst)}')
    print(f'2. n = {n} and results = {measure_time(u2_duplicated, numbers_worst)}')

```
